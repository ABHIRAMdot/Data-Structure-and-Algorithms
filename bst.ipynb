```python
class BST:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

    def insertBST(self, data): 
        if self.key is None:
            self.key = data
            return 

        if self.key == data:  # to avoid duplicate. otherwise it will insert to left or right
            return
        
        if data < self.key:
            if self.left:
                self.left.insertBST(data)
            else:
                self.left = BST(data)
        else:
            if self.right:
                self.right.insertBST(data)
            else:
                self.right = BST(data)


    def search(self,data):
        if data ==self.key:
            print("node is found")
            return True
        
        if data < self.key:
            if self.left:
                return self.left.search(data)
            else:
                print("node not present in tree")
                return False
        else:
            if self.right:
                return self.right.search(data)
            else:
                print("node not present in tree")
                return  False

    def preorder(self):
        print(self.key, end=" ")
        if self.left:
            self.left.preorder()
        if self.right:
            self.right.preorder()

    
    def inorder(self):
        if self.left:
            self.left.inorder()

        print(self.key, end=" ")
        if self.right:
            self.right.inorder()


    def postorder(self):
        if self.left:
            self.left.postorder()
        if self.right:
            self.right.postorder()
        print(self.key, end=" ")    

    # find minimum (helper fuction)
    def find_min(self):
        current = self
        while current.left:
            current = current.left
        
        return current


    def delete(self, data):
        
        if data < self.key:
            if self.left:
               self.left = self.left.delete(data)   # assign the value that return from the recursion to self.left
            else:
                print("Node not Found")

        elif data > self.key:
            if self.right:
               self.right = self.right.delete(data)  # assign the value that return from the recursion to self.right
            else:
                print("Node not found")
        
        else:
            # Node found

            #case 1: no child
            if self.left is None and self.right is None:
                return None
            
            #case2 : One child
            if self.left is None:
                return self.right
            
            if self.right is None:
                return self.left
            
            # case3 : Two child
            node = self.right.find_min()
            self.key = node.key      # changing the value of the node to be deleted with min value
            self.right = self.right.delete(node.key)  # deleting the changed min value  node

        # after the if or elif condition gets their left or right node(after deletion) then it return the non deleted node 
        return self     
        


root=BST(21)
value =[10,30,5,3,3,12,25,100,3,7]
for i in value:
    root.insertBST(i)

# root.search(5)
print("preorder traversal")
root.preorder()
# print("\ninorder traversal")
# root.inorder()
# print("\npostorder traversal")
# root.postorder()

root.delete(12)
print()
root.preorder()
                    
```

Check is BST

```python
def is_bst(node, min_val=float('-inf'), max_val=float('inf')):
    if node is None:
        return True
    
    if not(min_val < node.key < max_val):
        return False
    
    return(
        is_bst(node.left, min_val, node.key) and is_bst(node.right, node.key, max_val)
    )
```

Check if Two BSTs are identical 

```python
def identical(t1, t2):
    if t1 is None and t2 is None:
        return True
    
    if t1 is None or t2 is None:
        return False
    
    if t1.key != t2.key:
        return False
    
    return (
        identical(t1.left, t2.left) and identical(t1.right, t2.right)

    )

```

Height of Tree

```python
def height(root):
    left_height = -1
    right_height = -1

    if root.left:
        left_height = height(root.left)

    if root.right:
        right_height = height(root.right)

    return max(left_height, right_height) + 1
 
```

Depth of node

```python
def depth_of_node(root, val, d=0):
    if root.key == val:
        return d
    
    if val < root.key and root.left:
        return depth_of_node(root.left, val, d+1)
    if val > root.key and root.right:
        return depth_of_node(root.right, val, d+1)
    
    return -1
```
