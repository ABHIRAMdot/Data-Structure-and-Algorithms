```python
class BST:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

    def insertBST(self, data): 
        if self.key is None:
            self.key = data
            return 

        if self.key == data:  # to avoid duplicate. otherwise it will insert to left or right
            return
        
        if data < self.key:
            if self.left:
                self.left.insertBST(data)
            else:
                self.left = BST(data)
        else:
            if self.right:
                self.right.insertBST(data)
            else:
                self.right = BST(data)


    def search(self,data):
        if data ==self.key:
            print("node is found")
            return True
        
        if data < self.key:
            if self.left:
                return self.left.search(data)
            else:
                print("node not present in tree")
                return False
        else:
            if self.right:
                return self.right.search(data)
            else:
                print("node not present in tree")
                return  False

    def preorder(self):
        print(self.key, end=" ")
        if self.left:
            self.left.preorder()
        if self.right:
            self.right.preorder()

    
    def inorder(self):
        if self.left:
            self.left.inorder()

        print(self.key, end=" ")
        if self.right:
            self.right.inorder()


    def postorder(self):
        if self.left:
            self.left.postorder()
        if self.right:
            self.right.postorder()
        print(self.key, end=" ")    

    # find minimum (helper fuction)
    def find_min(self):
        current = self
        while current.left:
            current = current.left
        
        return current


    def delete(self, data):
        
        if data < self.key:
            if self.left:
               self.left = self.left.delete(data)   # assign the value that return from the recursion to self.left
            else:
                print("Node not Found")

        elif data > self.key:
            if self.right:
               self.right = self.right.delete(data)  # assign the value that return from the recursion to self.right
            else:
                print("Node not found")
        
        else:
            # Node found

            #case 1: no child
            if self.left is None and self.right is None:
                return None
            
            #case2 : One child
            if self.left is None:
                return self.right
            
            if self.right is None:
                return self.left
            
            # case3 : Two child
            node = self.right.find_min()
            self.key = node.key      # changing the value of the node to be deleted with min value
            self.right = self.right.delete(node.key)  # deleting the changed min value  node

        # after the if or elif condition gets their left or right node(after deletion) then it return the non deleted node 
        return self     
        


root=BST(21)
value =[10,30,5,3,3,12,25,100,3,7]
for i in value:
    root.insertBST(i)

# root.search(5)
print("preorder traversal")
root.preorder()
# print("\ninorder traversal")
# root.inorder()
# print("\npostorder traversal")
# root.postorder()

root.delete(12)
print()
root.preorder()
                    
```

```python
class BST:
    def __init__(self,key):
        self.key =key 
        self.lchild=None
        self.rchild=None

    def insertBST(self,data):
        if self.key is None:
            self.key=data
            return
        # if self.key==data:#to avoid duplicate. Otherwise it will inserted to left or right subtree
        #     return
        if self.key>data:
            if self.lchild:
                self.lchild.insertBST(data)
            else:
                self.lchild=BST(data)
        else:
            if self.rchild:
                self.rchild.insertBST(data)
            else:
                self.rchild=BST(data)

    def search(self,data):
        if self.key==data:
            print("node is found..")
            return
        if data<self.key:
            if self.lchild:
                self.lchild.search(data)
            else:
                print("node is not present in tree")
        else:
            if self.rchild:
                self.rchild.search(data)
            else:
                print("node is not present in tree")

    def preorder(self):
        print(self.key,end=" ")
        if self.lchild:
            self.lchild.preorder()
        if self.rchild:
            self.rchild.preorder()
    
    def inorder(self):
        if self.lchild:
            self.lchild.inorder()
        print(self.key,end=" ")
        if self.rchild:
            self.rchild.inorder()

    def postorder(self):
        if self.lchild:
            self.lchild.postorder()
        if self.rchild:
            self.rchild.postorder()
        print(self.key,end=" ")

    def delete(self,data):
        # if self.key==None:
        #     print("Tree is empty")
        #     return
        if data<self.key:
            if self.lchild:
                self.lchild=self.lchild.delete(data)
            else:
                print("given node is not present")
        elif data>self.key:
            if self.rchild:
                self.rchild=self.rchild.delete(data)
            else:
                print("given node is not present")
        else :
            # Node Found
            if self.lchild is None:
                temp=self.rchild
                self=None
                return temp
            if self.rchild is None:
                temp=self.lchild
                self=None
                return temp
            #now we are finding smallest key in right subtree.to delete node with two child node
            node=self.rchild
            while node.lchild:
                node=node.lchild
            self.key=node.key
            self.rchild=self.rchild.delete(node.key)
        return self
            
        
        



root=BST(21)
values =[10,30,5,3,3,12,25,100,3,7]
for i in values:
    root.insertBST(i)

# root.search(5)
print("preorder traversal")
root.preorder()
# print("\ninorder traversal")
# root.inorder()
# print("\npostorder traversal")
# root.postorder()

root.delete(12)
print()
root.preorder()
```
